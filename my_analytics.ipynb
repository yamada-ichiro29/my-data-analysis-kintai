```python
import pandas as pd

data = {
    "商品名":["リンゴ","バナナ","メロン"],
    "価格":[150,100,500]
}
df = pd.DataFrame(data)
df 
```

```python
# 1.csvファイルの読み込み
calendar_df = pd.read_csv(r'C:\勤怠テーブル一覧\calendar_table.csv')
calendar_df

attendance_df = pd.read_csv(r'C:\勤怠テーブル一覧\attendance_table.csv')
attendance_df

employee_df = pd.read_csv(r'C:\勤怠テーブル一覧\employee_table.csv')
employee_df

department_df = pd.read_csv(r'C:\勤怠テーブル一覧\department_table.csv')
department_df

# 2.日時型に変換する
attendance_df['出勤時刻'] = pd.to_datetime(attendance_df['出勤時刻'])
attendance_df['退勤時刻'] = pd.to_datetime(attendance_df['退勤時刻'])
attendance_df['休憩時刻'] = pd.to_timedelta(attendance_df['休憩時刻'] + ':00')


# 3.実績データの重複防止
sub_att_df = attendance_df.groupby(['従業員id','年月日']).agg(
    最早出勤時刻 = ('出勤時刻','min')
    最遅出勤時刻 = ('退勤時刻','max')
    休憩時刻=('休憩時刻','max')
    
)
 

# 3.従業員テーブルと部署テーブルを結合
emp_dap_df = pd.merge(employee_df[['従業員id', '姓', '名','部署ID']],department_df[['部署ID', '部署名']], on='部署ID', how='left')
emp_dap_df['部署名'] = emp_dap_df['部署名'].fillna('未所属')
emp_dap_df['部署ID'] = emp_dap_df['部署ID'].fillna('000')
emp_dap_df

# 4.3のデータフレームとカレンダーを結合
cal_emp_df = pd.merge(calendar_df,emp_dap_df,how = 'cross')
cal_emp_df


```
